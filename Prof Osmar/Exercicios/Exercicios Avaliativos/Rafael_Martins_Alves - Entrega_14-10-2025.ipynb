```python
print("Menu de opções:")
print("1. Multiplicar dois números.")
print("2. Dividir dois números.")

opcao = int(input("Digite a opção desejada do menu (1 ou 2): "))

if opcao == 1:
    mult_1 = float(input("Digite o primeiro número da multiplicação: "))
    mult_2 = float(input("Digite o segundo número da multiplicação: "))
    
    total = mult_1 * mult_2
    print("Este é o total do seu cálculo:", total)
    
elif opcao == 2:
    div_1 = float(input("Digite o primeiro número da divisão: "))
    div_2 = float(input("Digite o segundo número da divisão: "))
    
    if div_2 == 0:
        print("Não existe divisão por zero. ")
    else:
        total = div_1 / div_2
        print("Este é o total do seu cálculo:", total)
else:
    print("Opção não disponível. ")
```

```python
cod_produto = int(input("Digite o código do produto (1 a 10): "))
kilos_produto = float(input("Digite o peso do produto em quilo: "))
cod_pais = int(input("Digite o código do país de origem do produto (1 a 3): "))

peso_em_gramas = kilos_produto * 1000

preco_por_grama = 0
if (cod_produto >= 1 and cod_produto <= 4):
    preco_por_grama = 10
elif (cod_produto >= 5 and cod_produto <= 7):
    preco_por_grama =  25
elif (cod_produto >= 8 and cod_produto <= 10):
    preco_por_grama = 35

preco_total_por_grama = preco_por_grama * peso_em_gramas

valor_imposto = 0
if (cod_pais == 1):
    valor_imposto = 0
elif (cod_pais == 2):
    valor_imposto = preco_total_por_grama * 0.15
elif (cod_pais == 3):
    valor_imposto = preco_total_por_grama * 0.25

preco_total_com_imposto = preco_total_por_grama + valor_imposto
        
print("O peso do produto convertido em gramas é:", peso_em_gramas, "gramas")
print("O preço total do produto calculado pelas gramas é: R$ %.2f" % preco_total_por_grama)
print("O valor que deverá ser pago em impostos pelo produto é: R$ %.2f" % valor_imposto)
print("O valor total do produto com impostos é: R$ %.2f" % preco_total_com_imposto)
```

```python
num = int(input("Digite um número inteiro: "))

contador = 2

if num <= contador:
    primo = True
else:
    while contador < num:
        primo = True
        
        if num < contador:
            primo = True
        elif num % contador == 0:
            primo = False
            break
        
        contador += 1
        
if primo:
    print("Este número é primo!")
else:
    print("Este número não é primo!")
```

```python
saque = int(input("Digite o valor que deseja sacar: "))
saque_total = saque
contador_nota_cem = 0
contador_nota_cinquenta = 0
contador_nota_vinte = 0
contador_nota_dez = 0

while saque < 10 or saque % 10 != 0:
    saque = int(input("O valor deve ser inteiro positivo e múltiplo de 10. Ex: 270: "))
    saque_total = saque
    
while saque >= 10:
    if saque >= 100:
        saque -= 100
        contador_nota_cem += 1
    elif saque >= 50:
        saque -= 50
        contador_nota_cinquenta += 1
    elif saque >= 20:
        saque -= 20
        contador_nota_vinte += 1
    elif saque >= 10:
        saque -= 10
        contador_nota_dez += 1

print("Saque realizado:", saque_total, "reais!")
print("Notas de 100:", contador_nota_cem)
print("Notas de 50:", contador_nota_cinquenta)
print("Notas de 20:", contador_nota_vinte)
print("Notas de 10:", contador_nota_dez)
```
